```python
# ==============================================================================
# SCRIPT 2: Cálculo de las Tasas de Disociación (k_off)
# ==============================================================================
# ESTE ES CON REPLICA 1
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# 1. Cargar los datos de disociación
df = pd.read_excel('DatosLimpiosFasedeDisociacion.xlsx')

# 2. Definir la ecuación de disociación (decaimiento exponencial)
def ecuacion_disociacion(t, Y0, Fondo, k_off):
    """
    t: Tiempo ajustado de disociación
    Y0: Fluorescencia inicial al empezar a lavar
    Fondo: Fluorescencia residual o línea base
    k_off: Constante de disociación
    """
    return (Y0 - Fondo) * np.exp(-k_off * t) + Fondo

# 3. Extraer las variables
tiempo_dis = df['t_adj_dis'].values
fluorescencia_dis = df['Replica_1'].values

# 4. Ajuste de curva (Curve Fitting)
# Suposiciones iniciales: Y0 ~ 630, Fondo ~ 200, k_off ~ 0.01
parametros_iniciales = [630, 200, 0.01]

popt, pcov = curve_fit(ecuacion_disociacion, tiempo_dis, fluorescencia_dis, p0=parametros_iniciales)

Y0_calc = popt[0]
Fondo_calc = popt[1]
k_off_calc = popt[2]

print("--- Resultados de Disociación ---")
print(f"Y0 (Fluorescencia inicial): {Y0_calc:.2f} au")
print(f"Fondo (Señal residual): {Fondo_calc:.2f} au")
print(f"k_off (Tasa de disociación): {k_off_calc:.4f} min^-1")

# 5. Visualización
plt.figure(figsize=(8, 5))
plt.scatter(tiempo_dis, fluorescencia_dis, color='purple', label='Datos Experimentales (Replica 1)')

tiempo_fino = np.linspace(min(tiempo_dis), max(tiempo_dis), 100)
fluorescencia_modelo = ecuacion_disociacion(tiempo_fino, Y0_calc, Fondo_calc, k_off_calc)

plt.plot(tiempo_fino, fluorescencia_modelo, color='orange', linestyle='--', label='Modelo Ajustado de Caída')

plt.title('Cinética de Disociación de la Sonda Fluorescente')
plt.xlabel('Tiempo de Disociación Ajustado (min)')
plt.ylabel('Intensidad de Fluorescencia (au)')
plt.legend()
plt.grid(False)
plt.show()
plt.rcParams["font.weight"] = "bold"
plt.rcParams["axes.labelweight"] = "bold"
plt.rcParams["axes.titleweight"] = "bold"
```
