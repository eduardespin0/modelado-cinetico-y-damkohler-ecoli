```python
# ==============================================================================
# SCRIPT 1: Cálculo de la Tasas Observadas de Asociación (k_obs)
# ==============================================================================

# ESTE ES CON REPLICA 1
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# 1. Cargar los datos
df = pd.read_excel('DatosLimpiosFasedeAsociacion.xlsx')

# 2. Definir la ecuación de asociación cinética
def ecuacion_asociacion(t, Y_max, k_obs):
    return Y_max * (1 - np.exp(-k_obs * t))

# 3. Extraer las variables
tiempo_ajustado = df['t_adj'].values
fluorescencia = df['Replica_1'].values

# 4. Ajuste de curva (Curve Fitting)
parametros_iniciales = [600, 0.05]
popt, pcov = curve_fit(ecuacion_asociacion, tiempo_ajustado, fluorescencia, p0=parametros_iniciales)

Y_max_calculado = popt[0]
k_obs_calculado = popt[1]

print("--- Resultados del Modelado Cinético ---")
print(f"Y_max (Fluorescencia en equilibrio): {Y_max_calculado:.2f} au")
print(f"k_obs (Tasa observada): {k_obs_calculado:.4f} min^-1")

# 5. Visualización para validar el modelo
plt.figure(figsize=(8, 5))
plt.scatter(tiempo_ajustado, fluorescencia, color='blue', label='Datos Experimentales (Replica 1)')

tiempo_fino = np.linspace(min(tiempo_ajustado), max(tiempo_ajustado), 100)
fluorescencia_modelo = ecuacion_asociacion(tiempo_fino, Y_max_calculado, k_obs_calculado)

plt.plot(tiempo_fino, fluorescencia_modelo, color='red', linestyle='--', label='Modelo Ajustado')

plt.title('Cinética de Asociación de la Sonda Fluorescente')
plt.xlabel('Tiempo Ajustado (min)')
plt.ylabel('Intensidad de Fluorescencia (au)')
plt.legend()
plt.grid(False)
plt.show()
plt.rcParams["font.weight"] = "bold"
plt.rcParams["axes.labelweight"] = "bold"
plt.rcParams["axes.titleweight"] = "bold"
```
