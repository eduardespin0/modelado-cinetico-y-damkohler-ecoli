```python
# ==============================================================================
# SCRIPT 3: Cálculo de Afinidad y Parámetros Cinéticos Finales
# ==============================================================================

# 1. Ingresar las tasas extraídas de los ajustes anteriores
k_obs = 0.04968   # Tasa de asociación observada (min^-1)
k_off = 0.04818   # Tasa de disociación (min^-1)

# 2. Definir la concentración del ligando (Sonda)
concentracion_uM = 200                # Concentración en micromolares
L_molar = concentracion_uM * 1e-6     # Conversión a Molar (0.0002 M)

# 3. Fórmulas Cinéticas
# Tasa de Asociación Real (k_on)
k_on = (k_obs - k_off) / L_molar

# Constante de Disociación de Equilibrio (K_D) - Mide afinidad (menor = más fuerte)
Kd_molar = k_off / k_on
Kd_mM = Kd_molar * 1000  # Convertir a milimolar para reporte

# Constante de Asociación (K_A) - Fuerza de retención
Ka = 1 / Kd_molar

# 4. Imprimir resultados estructurados
print("==========================================")
print("   PARÁMETROS CINÉTICOS DEFINITIVOS")
print("==========================================")
print(f"[L] Concentración: {concentracion_uM} uM ({L_molar} M)")
print(f"k_on (Tasa real):  {k_on:.1f} M^-1 min^-1")
print(f"K_D  (Afinidad):   {Kd_mM:.3f} mM")
print(f"K_A  (Fuerza):     {Ka:.2f} M^-1")
print("==========================================")
```
